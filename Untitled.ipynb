```python
import re
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
train = (
    pd.read_csv("../Datasets/train.csv")
    .rename(columns={"market_forward_excess_returns": "target"})
)

# Cast all columns except 'date_id' to float
cols_to_cast = [c for c in train.columns if c != "date_id"]
train[cols_to_cast] = train[cols_to_cast].astype(float)

# Drop last 10 rows
train = train.head(-10)
train
```

```python
def sort_cols(df, prefix):
    cols = [c for c in df.columns if c.startswith(prefix)]
    return sorted(cols, key=lambda x: int(re.search(r'\d+', x).group()) if re.search(r'\d+', x) else float('inf'))

M_cols = sort_cols(train, 'M')
E_cols = sort_cols(train, 'E')
I_cols = sort_cols(train, 'I')
P_cols = sort_cols(train, 'P')
V_cols = sort_cols(train, 'V')
S_cols = sort_cols(train, 'S')
D_cols = sort_cols(train, 'D')
```

## Trend

```python
def plot_trend(train, cols, window=504):
    """
    Plots each feature in a group with its 504-day moving average
    """
    n = len(cols)
    ncols = 3
    nrows = (n + ncols - 1) // ncols

    fig, axes = plt.subplots(nrows, ncols, figsize=(5 * ncols, 3.5 * nrows))
    axes = axes.flatten()

    for ax, col in zip(axes, cols):
        s = train[col]
        s.plot(ax=ax, style='.', color='0.6')
        s.rolling(window, center=True, min_periods=window//2).mean().plot(ax=ax, linewidth=2)
        ax.set_title(col)

    # Hide unused axes
    for ax in axes[len(cols):]:
        ax.set_visible(False)

    plt.tight_layout()
    plt.show()
```

```python
plot_trend(train, M_cols)
plot_trend(train, E_cols)
plot_trend(train, I_cols)
plot_trend(train, P_cols)
plot_trend(train, V_cols)
plot_trend(train, S_cols)
```

```python
target = train['target']
moving_average = target.rolling(window=252, center=True, min_periods=126).mean()

ax = target.plot(style='.', color='0.5', figsize=(10, 5))
moving_average.plot(
    ax=ax, linewidth=3, title='252-Day Moving Average of Target', legend=False
)
plt.show()
```

```python
print('\nAnnualized volatility of target: {}'.format((target.std() * np.sqrt(252)).round(5)))
```

## Seasonality

```python
X = train.copy()

X['week_id'] = X['date_id'] // 5  # week index
X['day_in_week'] = X['date_id'] % 5  # 0-4, position inside week
X['week_in_year'] = X['week_id'] % 51 # month position within each trading year (resets every 51 weeks)

X['month_id'] = X['date_id'] // 21  # month index
X['day_in_month'] = X['date_id'] % 21  # 0-20, position inside month
X['month_in_year'] = X['month_id'] % 12  # month position within each trading year (resets every 12 months)

X['quarter_id'] = X['date_id'] // 63
X['day_in_quarter'] = X['date_id'] % 63
X['quarter_in_year'] = X['quarter_id'] % 4

X['year_id'] = X['date_id'] // 252
X['day_in_year'] = X['date_id'] % 252
```

### Weekly
There are hints of weekly and annual-weekly seasonality.  
But patterns are weak, noisy, and not persistent across years, suggesting they’re regime-driven

```python
# Set consistent style
sns.set_theme(style="whitegrid")

# Create 2×2 subplot layout
fig, axes = plt.subplots(2, 2, figsize=(18, 10))

# Intra-week seasonality (average by weekday)
sns.lineplot(
    x='day_in_week',
    y='target',
    data=X.groupby('day_in_week', as_index=False)['target'].mean(),
    ax=axes[0, 0],
    marker='o',
)
axes[0, 0].set_title("Intra-week Seasonality (Day of Week)", fontsize=12)
axes[0, 0].set_xlabel("Day in Week (0 = Mon, 4 = Fri)")
axes[0, 0].set_ylabel("Mean Target")

# Stability heatmap (weekday pattern per year)
pivot = X.pivot_table(values='target', index='year_id', columns='day_in_week', aggfunc='mean')
sns.heatmap(
    pivot,
    cmap='coolwarm',
    center=0,
    ax=axes[0, 1],
)
axes[0, 1].set_title("Weekday Pattern Per Year", fontsize=12)
axes[0, 1].set_xlabel("Day in Week")
axes[0, 1].set_ylabel("Year ID")

# Inter-week (annual weekly) seasonality
sns.lineplot(
    x='week_in_year',
    y='target',
    data=X.groupby('week_in_year', as_index=False)['target'].mean(),
    ax=axes[1, 0],
    marker='o',
)
axes[1, 0].set_title("Inter-week (Annual Weekly) Seasonality", fontsize=12)
axes[1, 0].set_xlabel("Week in Year (0-50)")
axes[1, 0].set_ylabel("Mean Target")

# Stability heatmap for inter-week pattern (per year)
pivot = X.pivot_table(values='target', index='year_id', columns='week_in_year', aggfunc='mean')
sns.heatmap(
    pivot,
    cmap='coolwarm',
    center=0,
    ax=axes[1, 1],
)
axes[1, 1].set_title("Weekly Pattern Per Year", fontsize=12)
axes[1, 1].set_xlabel("Week in Year (0-50)")
axes[1, 1].set_ylabel("Year ID")

# Adjust layout and spacing
plt.tight_layout()
plt.show()
```
